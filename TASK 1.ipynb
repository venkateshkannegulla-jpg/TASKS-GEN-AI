2.4 Wrap it in a reusable function

```python
def ask_llm(user_prompt, system_prompt=None, temperature=0.7, model=MODEL):

    """Send a single-turn chat request to Groq and return the reply text."""

    messages = []

    if system_prompt:

        messages.append({"role": "system", "content": system_prompt})

    messages.append({"role": "user", "content": user_prompt})


    response = client.chat.completions.create(

        model=model,

        messages=messages,

        temperature=temperature,

    )

    return response


result = ask_llm("Give me a fun fact about Python the language.")

print(ask_llm.__doc__)

print(result.choices[0].message.content)
```

2.5 Experiment: Looking at token usage

```python
result = ask_llm("Explain what a token is, in exactly two sentences.")


print("Reply:\n", result.choices[0].message.content)

print("\nToken usage:")

print("  Prompt tokens:", result.usage.prompt_tokens)

print("  Completion tokens:", result.usage.completion_tokens)

print("  Total tokens:", result.usage.total_tokens)
```

```python
import os

import json

from groq import Groq

from google.colab import userdata


client = Groq(api_key=userdata.get("FSD"))

MODEL  = "openai/gpt-oss-120b"


print(" Groq client ready")
```

 1.Zero-Shot Prompting

```python
# ── Helper: single-turn call ──────────────────────────────────────────────────

def ask(prompt: str, system: str = "You are a helpful assistant.") -> str:

    resp = client.chat.completions.create(

        model=MODEL,

        max_tokens=512,

        messages=[

            {"role": "system", "content": system},

            {"role": "user",   "content": prompt}

        ],

    )

    return resp.choices[0].message.content



# ── Zero-shot sentiment classification ───────────────────────────────────────

zero_shot_prompt = """

Classify the sentiment of the following review as Positive, Negative, Neutral.

Reply with only the label.


Review: I really enjoyed the meeting.

Sentiment:"""


#Great! Another three-hour meeting. Just what I needed.


result = ask(zero_shot_prompt)

print(f"Model answer: {result}")



# ── Zero-shot translation ─────────────────────────────────────────────────────

translate_prompt = "Translate the following English sentence to india.\n\nSentence: Hello ..How are you!"

print(ask(translate_prompt))

```

2 · Few-Shot Prompting

```python
# ── Few-shot sentiment with custom labels ─────────────────────────────────────

few_shot_prompt = """

Classify the sentiment of each review. Use exactly one label: [BUY] [PASS] [SKIP]


Review: "Absolutely brilliant, exceeded every expectation!"

Label: [BUY]


Review: "Broke after two days. Total waste of money."

Label: [SKIP]


Review: "It's fine. Does what it says on the box."

Label: [PASS]


Review: "Life-changing product, would recommend to everyone."

Label:"""


result = ask(few_shot_prompt)

print(f"Predicted label: {result}")


# ── Few-shot sentiment with custom labels ─────────────────────────────────────


prompt = """

Classify each query as [ACADEMIC], [PLACEMENT], or [ADMIN].


Query: "When is the Python exam?"

Category: [ACADEMIC]


Query: "Which companies are coming for placements?"

Category: [PLACEMENT]


Query: "Where do I apply for a bonafide certificate?"

Category: [ADMIN]


Query: "When will the Python lab exam be conducted?"

Category:

"""

result = ask(prompt)

print(f"Predicted label: {result}")

### The LLM uses the relationships and patterns in the text it has learned during training, represented internally through its neural network.


# ── Few-shot named-entity extraction ─────────────────────────────────────────

ner_prompt = """

Extract the company and role from each sentence.


Sentence: "Sundar Pichai joined Google in 2004 as a product manager."

Company: Google | Role: Product Manager


Sentence: "Jensen Huang co-founded Nvidia and serves as its CEO."

Company: Nvidia | Role: CEO


Sentence: "Dario Amodei left OpenAI to start Anthropic as its chief executive."

Company:"""


print(ask(ner_prompt))

```

3 · Chain-of-Thought (CoT) Prompting

```python
# ── Zero-shot CoT ─────────────────────────────────────────────────────────────

cot_prompt = """

A store sells apples at ₹12 each and oranges at ₹8 each.

Ravi buys 5 apples and 3 oranges and pays with a ₹100 note.

How much change does he receive?


Let's think step by step.

"""


answer = ask(cot_prompt)

print(answer)



# ── Few-shot CoT: logical reasoning ──────────────────────────────────────────

few_shot_cot = """

Q: All roses are flowers. Some flowers fade quickly. Do all roses fade quickly?

A: Step 1 — Roses ⊆ Flowers.

   Step 2 — "Some flowers fade" means at least one does, not necessarily all.

   Step 3 — We cannot conclude ALL roses fade quickly.

   Answer: No.


Q: All mammals are warm-blooded. Dolphins are mammals. Are dolphins warm-blooded?

A:"""


print(ask(few_shot_cot))
```
